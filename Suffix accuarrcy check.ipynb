```python
import pandas as pd
import re
from collections import defaultdict

df = pd.read_csv("/content/drive/MyDrive/vassista22_code_switching_adalora_checkpoints_phase1/Correct_Root.csv")
```

```python
def split_root_suffix(token):
    m = re.match(r'^([A-Za-z]+)(.*)$', token)
    if m:
        root = m.group(1)
        suffix = m.group(2)
        if suffix == "":
            suffix = "∅"
        return root, suffix
    return None, None
```

```python
def build_map(sentence):
    mp = {}
    for tok in sentence.strip().split():
        root, suf = split_root_suffix(tok)
        if root:
            mp[root] = suf
    return mp
```

```python
insertions = 0
deletions = 0
substitutions = 0
correct = 0
total_roots = 0

for _, row in df.iterrows():
    R = build_map(row["Reference_Text"])
    H = build_map(row["Generated_Text"])

    for w in R:
        sR = R.get(w, "∅")
        sH = H.get(w, "∅")
        total_roots += 1

        if sR == sH:
            correct += 1
        elif sR != "∅" and sH == "∅":
            deletions += 1
        elif sR == "∅" and sH != "∅":
            insertions += 1
        else:
            substitutions += 1
```

```python
MSER = (insertions + deletions + substitutions) / total_roots

print("Total English roots:", total_roots)
print("Correct:", correct)
print("Insertions:", insertions)
print("Deletions:", deletions)
print("Substitutions:", substitutions)
print("MSER:", MSER)
print("Suffix Accuracy:", 1 - MSER)
```

```python
conf = defaultdict(lambda: defaultdict(int))

for _, row in df.iterrows():
    R = build_map(row["Reference_Text"])
    H = build_map(row["Generated_Text"])

    for w in R:
        sR = R.get(w, "∅")
        sH = H.get(w, "∅")
        conf[sR][sH] += 1

pd.DataFrame(conf).fillna(0).astype(int)
```
