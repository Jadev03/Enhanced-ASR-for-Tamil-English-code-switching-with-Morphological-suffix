```python
import pandas as pd
import re
from jiwer import wer,process_words
```

```python
TAMIL_REGEX = re.compile(r'^[\u0B80-\u0BFF]+$')

def is_pure_tamil_word(word):
    """
    Returns True only if the word is fully Tamil
    Example:
    ✅ பண்ணனும்
    ❌ Alibaba-க்கு
    ❌ tax
    """
    return bool(TAMIL_REGEX.match(word))
```

```python
def extract_tamil_only_sentence(text):
    words = text.split()
    tamil_words = [w for w in words if is_pure_tamil_word(w)]
    return " ".join(tamil_words)
```

```python
df = pd.read_csv("/content/drive/MyDrive/vassista22_code_switching_adalora_checkpoints_phase1/final_predictions.csv")
```

```python
tamil_wer_scores = []

for idx, row in df.iterrows():
    ref_tamil = extract_tamil_only_sentence(row["Reference_Text"])
    hyp_tamil = extract_tamil_only_sentence(row["Generated_Text"])

    # Skip if no Tamil words exist
    if len(ref_tamil.strip()) == 0:
        continue

    sentence_wer = wer(ref_tamil, hyp_tamil)
    tamil_wer_scores.append(sentence_wer)
```

```python
total_S = 0
total_D = 0
total_I = 0
total_N = 0

for idx, row in df.iterrows():
    ref_tamil = extract_tamil_only_sentence(row["Reference_Text"])
    hyp_tamil = extract_tamil_only_sentence(row["Generated_Text"])

    if len(ref_tamil.strip()) == 0:
        continue

    result = process_words(ref_tamil, hyp_tamil)

    total_S += result.substitutions
    total_D += result.deletions
    total_I += result.insertions
    total_N += result.hits + result.substitutions + result.deletions
```

```python
tamil_wer = (total_S + total_D + total_I) / total_N
print(f"Tamil–Tamil WER (global): {tamil_wer:.4f}")
```
